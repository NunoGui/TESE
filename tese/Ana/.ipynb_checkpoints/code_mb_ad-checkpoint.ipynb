```python
## datasets.py ##

import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler

def load_csv(file_path):
    try:
        # Load the CSV with the first row as headers
        df = pd.read_csv(file_path, delimiter=';')
        
        # Print column names
        print("Columns found in the CSV:")
        for col in df.columns:
            print(f"- {col}")

        return df
    except FileNotFoundError:
        print("Error: File not found.")
    except Exception as e:
        print(f"An error occurred: {e}")


def transform_ratings(dataset):
    
    dataset = dataset[['id_survey', 'id_photo', 'like_bool']]
    dataset = dataset.rename(columns={'id_survey': 'user', 'id_photo': 'item','like_bool': 'rating' })

    return dataset

def transform_ratings_emotions(dataset):
    
    dataset = dataset[['id_survey', 'id_photo', 'like_bool', 'anger', 'fear', 'disgust', 'sadness', 'happiness', 'surprise', 'neutral'  ]]
    dataset = dataset.rename(columns={'id_survey': 'user', 'id_photo': 'item','like_bool': 'rating' })

    dataset = stadardize(dataset, ['anger', 'fear', 'disgust', 'sadness', 'happiness', 'surprise', 'neutral' ])
    return dataset

def transform_ratings_vad(dataset):
    
    dataset = dataset[['id_survey', 'id_photo', 'like_bool', 'valence', 'arousal', 'dominance' ]]
    dataset = dataset.rename(columns={'id_survey': 'user', 'id_photo': 'item','like_bool': 'rating' })
    dataset = stadardize(dataset, ['valence', 'arousal', 'dominance'])
    return dataset

def transform_ratings_all(dataset):
    
    #dataset = dataset[['id_survey', 'id_photo', 'like_bool', 'valence', 'arousal', 'dominance' ]]
    dataset = dataset.rename(columns={'id_survey': 'user', 'id_photo': 'item','like_bool': 'rating' })
    #dataset = stadardize(dataset, ['anger', 'fear', 'disgust', 'sadness', 'happiness', 'surprise', 'neutral', 'valence', 'arousal', 'dominance' ])
    return dataset


def generate_dummy_similarity_data(ids, similarity_range=(0.1, 1.0)):
    # Create all possible pairs (item1, item2) where item1 != item2
    item_pairs = []
    for i in range(len(ids)):
        for j in range(i + 1, len(ids)):
            item_pairs.append((ids[i], ids[j]))
    
    # Generate random similarities for each pair in the specified range
    similarities = np.random.uniform(similarity_range[0], similarity_range[1], len(item_pairs))
    
    # Create the DataFrame
    similarity_data = pd.DataFrame(item_pairs, columns=['item1', 'item2'])
    similarity_data['sim'] = similarities
    
    # Return the DataFrame
    return similarity_data


def stadardize(df, list_of_columns):
    scaler = MinMaxScaler()

    cols_to_standardize = list_of_columns

    # Fit & transform only selected columns
    df[cols_to_standardize] = scaler.fit_transform(df[cols_to_standardize])

    return df
```

```python

## Split.py ##
from sklearn.model_selection import train_test_split
import pandas as pd

def split_user_data(data):
    train_list = []
    test_list = []

    for user_id, user_data in data.groupby('user'):
        if len(user_data) == 1:  # Edge case: only one item
            train_list.append(user_data)
        else:
            train, test = train_test_split(user_data, test_size=0.2, random_state=42)
            train_list.append(train)
            test_list.append(test)

    train_data = pd.concat(train_list).reset_index(drop=True)
    test_data = pd.concat(test_list).reset_index(drop=True)

    return train_data, test_data


import pandas as pd
from sklearn.model_selection import train_test_split
import numpy as np


def remove_users_all_zero_ratings(df):

    # Find users where all ratings == 0
    users_all_zeros = df.groupby('user')['rating'].transform(lambda x: (x == 0).all())
    users_all_ones = df.groupby('user')['rating'].transform(lambda x: (x == 1).all())

    # Keep only rows where not all ratings are 0
    df_filtered = df[~users_all_zeros].copy()
    df_filtered2 = df_filtered[~users_all_ones].copy()

    return df_filtered2

def split_user_data_balanced(data, test_size=0.2, random_state=42, max_test_items=3):
    
    data = remove_users_all_zero_ratings(data)
   
    train_list = []
    test_list = []

    for user_id, user_data in data.groupby('user'):
        if len(user_data) == 1:  # Edge case: only one item
            train_list.append(user_data)
        else:
            train, test = train_test_split(user_data, test_size=0.2, random_state=42)

            if (test['rating'] == 0).all():
         
                # --- Step 2: Randomly select one row from train_df where rating == 1
                candidate = train[train['rating'] == 1].sample(1, random_state=np.random.randint(10000))
                
                # --- Step 3: Randomly select an index in test_df to replace
                replace_idx = np.random.choice(test.index)
                
                # --- Step 4: Replace that row in test_df with the sampled one
                test.loc[replace_idx] = candidate.iloc[0]

            elif (test['rating'] == 1).all():
     
                # All are one → take one '0' from train_df
                candidate = train[train['rating'] == 0].sample(1, random_state=np.random.randint(10000))
                replace_idx = np.random.choice(test.index)
                test.loc[replace_idx] = candidate.iloc[0]


            train_list.append(train)
            test_list.append(test)

    train_data = pd.concat(train_list).reset_index(drop=True)
    test_data = pd.concat(test_list).reset_index(drop=True)


    return train_data, test_data
```

```python
## Similarities.py ##
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity


def cosine_sim(df):
    # Pivot: rows = users, columns = items, values = likes
    user_item_matrix = df.pivot_table(index="user", 
                                    columns="item", 
                                    values="rating", 
                                    fill_value=0)

    # --- Cosine similarity ---
    cos_sim = cosine_similarity(user_item_matrix)
    cosine_df = pd.DataFrame(cos_sim, 
                            index=user_item_matrix.index, 
                            columns=user_item_matrix.index)

    return cosine_df


def pearson_similarity(df):
    # Pivot into user-item matrix
    user_item_matrix = df.pivot_table(index="user",
                                      columns="item",
                                      values="rating",
                                      fill_value=0)

    # Compute Pearson correlation between users
    pearson_corr = user_item_matrix.T.corr(method='pearson')
    sim_df = pearson_corr.fillna(0)  # fill NaNs with 0 if no overlap

    return sim_df


def cosine_sim_vector(df):
    item_cols = df.columns[3:]
    df_pivot = df.pivot(index='user', columns='item', values=item_cols.tolist())

    # Flatten multiindex columns
    df_pivot.columns = [f"{item}_{feat}" for item, feat in df_pivot.columns]

    # Fill missing values with 0 (if some users don't have vectors for some items)
    df_pivot = df_pivot.fillna(0)

    # Compute cosine similarity between users
    similarity_matrix = cosine_similarity(df_pivot)
    similarity_df = pd.DataFrame(similarity_matrix, index=df_pivot.index, columns=df_pivot.index)


    return similarity_df



def get_k_most_sim_users(similarity_df,target_user, k ):

    if target_user not in similarity_df.index:
        raise ValueError(f"User {target_user} not found in dataset.")

    # Get similarities for the target user, drop self
    sim_scores = similarity_df.loc[target_user].drop(target_user)

    # Return top-k users
    return sim_scores.nlargest(k)


def get_scores(test_items, similar_users, df_train):
    # Compute weighted scores for each test item
        # Pivot into user-item matrix
    user_item_matrix = df_train.pivot_table(index="user", 
                                      columns="item", 
                                      values="rating", 
                                      fill_value=0)
    
   # Compute mean score for each test item
    neighbor_ids = similar_users.index
    scores = {}
    for item in test_items:
        if item in user_item_matrix.columns:
            neighbor_ratings = user_item_matrix.loc[neighbor_ids, item]
            scores[item] = neighbor_ratings.mean()  # just take the mean
        else:
            scores[item] = 0.0  # unseen item

    return pd.Series(scores).sort_values(ascending=False)
```

```python
## evaluation.py ##

def precision_at_k(test_items, recommended_items, k):
    """
    Calculate Precision@k.
    
    Parameters:
    - test_items: list of items that are actually relevant (ground truth)
    - recommended_items: list of items recommended by the system
    - k: number of top recommendations to consider
    
    Returns:
    - precision: fraction of recommended items in top-k that are relevant
    """
    if not recommended_items or k == 0:
        return 0.0
    
    recommended_at_k = recommended_items[:k]
    relevant_items = set(test_items)
    
    # Count how many recommended items are actually relevant
    num_relevant = sum([1 for item in recommended_at_k if item in relevant_items])
    
    precision = num_relevant / k
    return precision



def precision_curve(recommended, relevant):
    """
    Calculate precision at every position in the recommendation list.
    
    Args:
        recommended (list): Ranked list of recommended items.
        relevant (set or list): Set of relevant (ground truth) items.
    
    Returns:
        list: Precision values at each position (1..len(recommended))
    """
    relevant = set(relevant)  # convert to set for faster lookup
    precisions = []
    hits = 0

    for i, item in enumerate(recommended, start=1):
        if item in relevant:
            hits += 1
        precisions.append(hits / i)

    return precisions



import matplotlib.pyplot as plt

def hit_ratio_curve_single(recommended_items, relevant_items, plot=False):
    """
    Compute the Hit Ratio curve (HR@k) for a single list of recommendations.
    
    Parameters:
    -----------
    recommended_items : list
        Ranked list of recommended item IDs.
    relevant_items : list
        List of ground-truth relevant item IDs.
    plot : bool
        Whether to plot the HR@k curve.
    
    Returns:
    --------
    ks : list of int
        The cutoff values (1..N)
    hr_values : list of float
        The Hit Ratio at each cutoff (fraction of relevant items found).
    """
    ks = range(1, len(recommended_items) + 1)
    hr_values = []

    hits = 0
    total_relevant = len(relevant_items)

    for k in ks:
        if recommended_items[k - 1] in relevant_items:
            hits += 1
        hr = hits / total_relevant  # fraction of relevant items retrieved so far
        hr_values.append(hr)

    if plot:
        plt.figure(figsize=(7, 4))
        plt.plot(ks, hr_values, marker='o')
        plt.title("Hit Ratio Curve (HR@k)")
        plt.xlabel("k (Top-k recommendations)")
        plt.ylabel("Hit Ratio")
        plt.grid(True)
        plt.show()
    print(recommended_items)
    print(relevant_items)
    print(list(ks), hr_values)
    return hr_values

def mean_reciprocal_rank_at_k(recommended_items, relevant_items):
    """
    Compute MRR@k for all k = 1..len(recommended_items)
    
    Parameters
    ----------
    recommended_items : list
        Ranked list of recommended item IDs.
    relevant_items : list
        List of relevant (ground truth) item IDs.
    
    Returns
    -------
    list
        List of MRR@k values for k = 1..len(recommended_items)
    """
    mrr_values = []
    found_rank = None
    
    # Iterate through each cutoff k
    for k in range(1, len(recommended_items) + 1):
        # Check if a relevant item appears within top-k
        for rank, item in enumerate(recommended_items[:k], start=1):
            if item in relevant_items:
                found_rank = rank
                break
        
        if found_rank is not None:
            mrr_values.append(1.0 / found_rank)
        else:
            mrr_values.append(0.0)
    
    return mrr_values


```

```python
## Main ##

import datasets
import numpy as np
import sys
import rs_cf_algorithms
import pandas as pd
import split

def main():

   
    ratings = datasets.load_csv("data/ratings_1025.csv")

    ratings = datasets.transform_ratings_all(ratings)

    precision_cf_likes = pd.DataFrame(0, index=range(1), columns=['top1', 'top2', 'top3'])
    mrr_cf_likes = pd.DataFrame(0, index=range(1), columns=['top1', 'top2', 'top3'])
    precision_cf_emotions = pd.DataFrame(0, index=range(1), columns=['top1', 'top2', 'top3'])
    mrr_cf_emotions = pd.DataFrame(0, index=range(1), columns=['top1', 'top2', 'top3'])
    precision_cf_vad = pd.DataFrame(0, index=range(1), columns=['top1', 'top2', 'top3'])
    mrr_cf_vad = pd.DataFrame(0, index=range(1), columns=['top1', 'top2', 'top3'])
    precision_cf_all = pd.DataFrame(0, index=range(1), columns=['top1', 'top2', 'top3'])
    mrr_cf_all = pd.DataFrame(0, index=range(1), columns=['top1', 'top2', 'top3'])
    precision_fc_random = pd.DataFrame(0, index=range(1), columns=['top1', 'top2', 'top3'])
    mrr_cf_random = pd.DataFrame(0, index=range(1), columns=['top1', 'top2', 'top3'])



    k_folds = 5
    n = 1 # most similar users
    for k in range(k_folds):
        train_data, test_data = split.split_user_data_balanced(ratings)
        print('k = ', k)
        precision_likes, mrr_likes = rs_cf_algorithms.CF_likes(train_data[['user', 'item', 'rating']], test_data[['user', 'item', 'rating']], n)
        precision_cf_likes += precision_likes
        mrr_cf_likes += mrr_likes

        precision_emotions, mrr_emotions = rs_cf_algorithms.CF_emotions(train_data[['user', 'item', 'rating','anger', 'fear', 'disgust', 'sadness', 'happiness', 'surprise', 'neutral']], test_data[['user', 'item', 'rating','anger', 'fear', 'disgust', 'sadness', 'happiness', 'surprise', 'neutral']], n)
        precision_cf_emotions += precision_emotions
        mrr_cf_emotions += mrr_emotions

        precision_vad, mrr_vad = rs_cf_algorithms.CF_vad(train_data[['user', 'item', 'rating','valence', 'arousal', 'dominance']], test_data[['user', 'item', 'rating','valence', 'arousal', 'dominance']], n)
        precision_cf_vad += precision_vad
        mrr_cf_vad += mrr_vad

        precision_all, mrr_all =rs_cf_algorithms.CF_all(train_data, test_data, n)
        precision_cf_all +=  precision_all
        mrr_cf_all +=  mrr_all


        precision_random, mrr_random =rs_cf_algorithms.CF_random(train_data, test_data, n)
        precision_fc_random +=  precision_random
        mrr_cf_random +=  mrr_random



    print('### ---- ###')


    final_precision_df = pd.concat([precision_cf_likes/k_folds, precision_cf_emotions/k_folds, precision_cf_vad/k_folds, precision_cf_all/k_folds, precision_fc_random/k_folds], ignore_index=True)
    print(final_precision_df)

    final_mrr_df = pd.concat([mrr_cf_likes/k_folds, mrr_cf_emotions/k_folds, mrr_cf_vad/k_folds, mrr_cf_all/k_folds, mrr_cf_random/k_folds], ignore_index=True)
    print(final_mrr_df)

    final_precision_df.to_csv('precision_all_n20_kfold5.csv', index=False)
    final_mrr_df.to_csv('mrr_all_n20_kfold5.csv', index=False)

if __name__ == "__main__":
    main()
```
